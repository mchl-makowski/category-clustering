# Service Desk Problem clustering: exploration

Thin wrapper over `problem_clustering`. Same YAML as the CLI. Defaults to the test fixture + Fake backend (offline); on Foundry set `INPUT` to the export and `OVERRIDES = {}`.

```python
from pathlib import Path

from problem_clustering import load_config, load_tickets, run_method, write_outputs
from problem_clustering.backend import make_backend

ROOT = Path.cwd() if (Path.cwd() / 'pyproject.toml').exists() else Path.cwd().parent
INPUT = ROOT / 'tests' / 'fixtures' / 'tickets.xlsx'
CONFIG = ROOT / 'config.example.yaml'
OVERRIDES = {'backend.kind': 'fake', 'method_a.distance_threshold': 0.6}
WRITE = False  # True: write yymmdd_hhmm_<method>.xlsx/.txt to ROOT / 'outputs'
```

```python
config = load_config(CONFIG, OVERRIDES)
tickets = load_tickets(INPUT)
backend = make_backend(config.backend)
result = run_method(tickets, 'a', config, backend)
print(result.log[-1])
```
